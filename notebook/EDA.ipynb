✔️ Churn Rate by Category

```python
df['Churn_num'] = df['Churn'].map({'Yes':1,'No':0})
```

```python
plt.figure(figsize=(6,4))
sns.barplot(data=df, x='gender', y='Churn_num')
plt.title("Churn Rate by Gender")


```
